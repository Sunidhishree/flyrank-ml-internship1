My lane is content clustering — grouping pages by shared behavioral patterns (traffic level, click-through rate, position stability, staleness) rather than predicting a single score or label. I chose this because most obvious lanes (decline prediction, opportunity ranking) reduce every page to one number, but in practice a content team doesn't manage pages one at a time — they manage them by type. A "high-traffic, low-CTR" page needs a different fix than a "stale but stable" page. Clustering lets me discover those natural groups directly from the data instead of assuming I already know the categories.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Clustering is unsupervised, so there is no target label to predict. Instead of a target, I'll define the feature space the clusters are built from: impressions_90d, ctr, avg_position (filtered to exclude the 0/no-data rows), and days_since_last_update. These are all observed values already in the dataset — not derived from a rule — so there's no label-leakage risk the way there would be with trend_direction/trend_pct.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

I'll use silhouette score as my quantitative metric — it measures how well-separated and internally consistent the clusters are (closer to 1 is better; near 0 means clusters overlap). Since silhouette alone can't tell me if clusters are meaningful to a content strategist, I'll pair it with a human sense-check: pulling a few example pages from each cluster and confirming they represent genuinely different, actionable page "types."

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule (e.g., "if impressions > X and CTR < Y, it's group A") only works along one or two dimensions at a time and forces me to guess the thresholds and boundaries in advance. Real pages vary across at least four interacting signals — traffic, CTR, position, and staleness — and the natural groupings likely don't fall along clean rectangular cutoffs. Clustering can find groups defined by the combination of these signals (e.g., "moderate traffic, low CTR, but recently updated" vs. "high traffic, high CTR, but stale") that I wouldn't think to hand-code as if/else rules.
