## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

My lane is content clustering — grouping pages by shared behavioral patterns (traffic level, click-through rate, position stability, staleness) rather than predicting a single score or label. I chose this because most obvious lanes (decline prediction, opportunity ranking) reduce every page to one number, but in practice a content team doesn't manage pages one at a time — they manage them by type. A "high-traffic, low-CTR" page needs a different fix than a "stale but stable" page. Clustering lets me discover those natural groups directly from the data instead of assuming I already know the categories.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

The decision this supports: which playbook/treatment to apply to a given page, and which cluster to prioritize first with limited team capacity. A content strategist would act on it — instead of reviewing pages one at a time, they'd pull up "cluster 3: high-impression, low-CTR pages" and apply a CTR-focused fix (better title/meta) to all of them at once. The cost of a wrong call: if pages get miscategorized, the team could waste effort applying the wrong fix — e.g., trying to improve CTR on a page whose real problem is low impressions/visibility, which no title tweak will fix.

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

My work will produce observed, descriptive groupings — patterns in traffic, CTR, and position that exist in this dataset over this time window. These are directional and decision-support: they suggest which pages resemble each other and might benefit from similar treatment. My work will never claim the clusters represent causal categories (e.g., "this cluster of pages performs poorly because of X"), and it will never claim to predict or reverse-engineer Google's ranking algorithm. Cluster assignments are a starting point for human review, not an automated verdict.
